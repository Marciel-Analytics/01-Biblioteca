Essa próxima função é pra ver se tem algum número nulo

aula 06 Filtrando os Dados I

```python
aluguel_casas = aluguel_casas[(aluguel_casas["total"]<=4000)]
aluguel_casas.head()
```

```python
aluguel_casas.describe()
```

agora vamos colocar uma premissa pro négocio/ algo que precisa ser respeitado / função "AND" 

tem que ser uma casa que aceite animais 

```python
aluguel_casas = aluguel_casas[(aluguel_casas["total"]<=4000) & (aluguel_casas["animal"]=="accept")]
aluguel_casas.head()
```

Agora vamos acrescentar um filtro de "OR" 

```python
aluguel_casas =  aluguel_casas[(aluguel_casas["total"]<=4000) & (aluguel_casas["animal"]=="accept") & ((aluguel_casas["rooms"]==2) | (aluguel_casas["rooms"]==3))]
aluguel_casas.head()
```

```python
aluguel_casas.describe()
```

Aula 07 Filtrando os Dados II

Filtrando dados utilizando uma função mais facil / função.query

```python
aluguel_casas = aluguel_casas.query("total<= 4000 and animal == 'accept'")
aluguel_casas.head()
```

```python
aluguel_casas = aluguel_casas.query(("total<= 4000 and animal == 'accept' and rooms == 2 or rooms == 3"))
aluguel_casas.head()
```

Aula 08 Agrupando os dados

```python
agg_city_price = aluguel_casas.groupby(["city"])["total"].agg("mean").reset_index()
agg_city_price
```

Agora vamos agrupar por quartos e por quantidades de banheiros 

```python
agg_bath_room_price =  aluguel_casas.groupby(["bathroom","rooms"])["total"].agg("mean").reset_index()
agg_bath_room_price
```

Aula 09 Visualizando os dados

```python
import seaborn as sns 
```

```python
ax = sns.barplot(data=agg_bath_room_price, x="rooms", y="total", color = "blue")
ax.set(xlabel = "qtd quartos", ylabel = "media do preço");
```

agora vai ser quantidade de banheiros 

```python
ax = sns.barplot(data=agg_bath_room_price, x="bathroom", y="total", color = "green")
ax.set(xlabel = "qtd banheiros", ylabel = "media do preço");
```
